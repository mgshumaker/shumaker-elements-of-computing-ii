```python
#Q10: Produce a list of facilities with a total revenue less than 1000.
#The output of facility name and total revenue, sorted by revenue.
#Remember that there's a different cost for guests and members!
query_total_revenue = """
                      SELECT SUM(
                        CASE
                            WHEN memid > 0 THEN membercost
                            WHEN memid = 0 THEN guestcost
                            ELSE "unknown"
                        END) as "total_revenue"
                      FROM facilities f
                      JOIN bookings b ON b.facid = f.facid
                      GROUP BY name
                      HAVING total_revenue < 1000
                      """

pd.read_sql(query_total_revenue, conn)
```

```python
# In-class code for Q11.

query_usage_by_in_class = """
                          SELECT f.name AS facility_name, m.firstname ||" "|| m.surname AS fullname, SUM(b.slots) AS usage
                          FROM bookings b
                          JOIN facilities f ON b.facid = f.facid
                          JOIN members m ON b.memid = m.memid
                          WHERE b.memid != 0
                          GROUP BY f.name, m.firstname, m.surname
                          ORDER BY m.surname, m.firstname;
                          """

pd.read_sql(query_usage_by_in_class, conn)
```

```python
#Q12: Find the facilities usage by month. Exclude guests
query_usage_ = """
               SELECT SUBSTRING(starttime, 6, 2) as month, f.name AS facility_name, SUM(b.slots) AS usage
               FROM facilities f 
               JOIN bookings b ON b.facid = f.facid
               WHERE b.memid != 0
               GROUP BY month, facility_name
               ORDER BY month, facility_name;
               """

pd.read_sql(query_usage_, conn)
```

# **1. Revenue & Financial KPIs**


**Total Revenue per Facility:**

Summing booking slots multiplied by facility costs (guest vs. member rates) to determine which facilities generate the most income.

```python
total_revenue_per_facility = """
                             SELECT f.name AS facility_name, SUM(
                             CASE
                                WHEN b.memid = 0 THEN f.guestcost * b.slots
                                ELSE f.membercost * b.slots
                             END) AS total_revenue
                             FROM bookings b
                             JOIN facilities f ON b.facid = f.facid
                             GROUP BY facility_name
                             ORDER BY total_revenue DESC;
                             """

pd.read_sql(total_revenue_per_facility, conn)
```

**Average Revenue per Member (ARPM):**

Calculate the total spending (fees + booking costs) per member to identify high-value users.

```python
average_revenue_per_member = """
                             SELECT m.firstname ||" "|| m.surname AS "Member Name", SUM(f.membercost * b.slots) AS "Total Spent"
                             FROM members m
                             JOIN bookings b ON b.memid = m.memid
                             JOIN facilities f on f.facid = b.facid
                             WHERE m.memid != 0
                             GROUP BY "Member Name"
                             ORDER BY "Total Spent" DESC
                             """

pd.read_sql(average_revenue_per_member, conn)
```

Total income generated specifically from non-member guests, indicating the popularity of the club with outside visitors.
